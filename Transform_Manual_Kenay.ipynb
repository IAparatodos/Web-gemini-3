## 🚀 Paso 3: Ejecutar Script de Transformación

```python
# Importar el script completo
!wget -q https://raw.githubusercontent.com/IAparatodos/Web-gemini-3/claude/document-transformation-script-01C7euYfsJibPxFTsvaxsVZJ/transform_manual_kenay.py

# Importar las funciones necesarias
from transform_manual_kenay import transform_manual
import os

# CONFIGURACIÓN
PROVIDER_PDF = "Instrucciones proveedor.pdf"  # 👈 Cambiar si tu archivo tiene otro nombre
REFERENCE_PDF = "Instrucciones Kenay (1).pdf"  # 👈 Opcional
OUTPUT_PDF = "Manual_Final_Kenay.pdf"

# Verificar archivos
if not os.path.exists(PROVIDER_PDF):
    print(f"❌ ERROR: No se encuentra '{PROVIDER_PDF}'")
    print("   Archivos disponibles:", os.listdir('.'))
    print("\n   Por favor, verifica el nombre del archivo en la celda anterior.")
else:
    # COORDENADAS PERSONALIZADAS (Ajustar según tu PDF)
    custom_coords = {
        0: (50, 200, 550, 750),   # Página 1 - área del diagrama
        1: (50, 200, 550, 750),   # Página 2
        2: (50, 200, 550, 750),   # Página 3
        3: (50, 200, 550, 750),   # Página 4
        4: (50, 200, 550, 750),   # Página 5
        5: (50, 200, 550, 750),   # Página 6
        6: (50, 200, 550, 750),   # Página 7
        7: (50, 200, 550, 750),   # Página 8
    }
    
    # EJECUTAR TRANSFORMACIÓN
    try:
        output_file = transform_manual(
            provider_pdf_path=PROVIDER_PDF,
            reference_pdf_path=REFERENCE_PDF if os.path.exists(REFERENCE_PDF) else None,
            output_pdf_path=OUTPUT_PDF,
            total_steps=8,
            custom_crop_coords=custom_coords
        )
        
        print("\n" + "="*60)
        print("🎉 ¡TRANSFORMACIÓN COMPLETADA!")
        print("="*60)
        print(f"\n📄 Archivo generado: {output_file}")
        print("\n✅ Continúa con el Paso 4 para descargar el PDF")
        
    except Exception as e:
        print(f"\n❌ ERROR: {str(e)}")
        print("\n💡 Sugerencias:")
        print("   1. Verifica que el PDF del proveedor sea válido")
        print("   2. Ajusta las coordenadas 'custom_coords' según tu PDF")
        print("   3. Revisa el error detallado arriba")
```

## 🔧 Configuración Avanzada (Opcional)

### Ajustar Coordenadas de Recorte

Si los diagramas no se extraen correctamente, necesitas ajustar las coordenadas `custom_coords`.

**Cómo encontrar las coordenadas correctas:**

```python
# HERRAMIENTA: Visualizar página del PDF para encontrar coordenadas
import fitz
from PIL import Image
import io

PROVIDER_PDF = "Instrucciones proveedor.pdf"
PAGE_NUM = 0  # 👈 Cambiar número de página (0 = primera página)

if os.path.exists(PROVIDER_PDF):
    doc = fitz.open(PROVIDER_PDF)
    page = doc[PAGE_NUM]
    
    # Renderizar página completa
    pix = page.get_pixmap(matrix=fitz.Matrix(2, 2))
    img = Image.open(io.BytesIO(pix.tobytes("png")))
    
    print(f"📄 Página {PAGE_NUM + 1}")
    print(f"📏 Dimensiones: {page.rect.width} x {page.rect.height} puntos")
    print("\n💡 Usa estas dimensiones para ajustar custom_coords")
    print("   Formato: (x0, y0, x1, y1)")
    print("   x0, y0 = esquina superior izquierda")
    print("   x1, y1 = esquina inferior derecha")
    
    display(img)
    doc.close()
else:
    print(f"❌ No se encuentra {PROVIDER_PDF}")
```
